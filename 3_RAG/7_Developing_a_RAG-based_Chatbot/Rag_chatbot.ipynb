```python
# We will use the faq.joblib file to create a FAQ Layout

# The FAQ Layout will be like this:

# Question: 'FAQ_Question_1', Answer: 'FAQ_Answer_1', Type: 'FAQ_Type_1'
...
# Question: FAQ Question 25, Answer: FAQ Answer 25, Type: FAQ Type 25
```

```python
def generate_faq_layout(faq_dict: list) -> str:
    
    t = ""

    for f in faq_dict:
        t += f"Question: {f['question']} Answer: {f['answer']} Type: {f['type']}\n" 
  

    return t
```
