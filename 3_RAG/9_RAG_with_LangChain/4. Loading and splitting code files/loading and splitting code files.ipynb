## 1. loading and splitting for markdown files

```python
# 1. Loading

from langchain_community.document_loaders import UnstructuredMarkdownLoader

markdown_loader = UnstructuredMarkdownLoader("sample-1.md")

documents = markdown_loader.load()

print(documents[0])
```

## 2. loading and splitting for python files

```python
# 1. Loading


from langchain_community.document_loaders import PythonLoader

python_loader = PythonLoader("example.py")

documents = python_loader.load()

print(documents[0])
```

```python
# 2. Splitting

from langchain_text_splitters import RecursiveCharacterTextSplitter

python_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 150,
    chunk_overlap= 10
)

chunks = python_splitter.split_documents(documents)

for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: \n{chunk.page_content}\n")
    print("-----"*10)
```

<i>class Anthropic(LLM):</i> is in chunk 2 , while it should be in chunk 3; 

That's why <b> there is a better method for splitting code files using <u> Language


```python
# Splitting by language

from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

python_splitter = RecursiveCharacterTextSplitter.from_language(
    language= Language.PYTHON, # ON MENTIONNE LE LANGAGE PYTHON!!!!!!!
    chunk_size = 150,
    chunk_overlap= 10)


chunks = python_splitter.split_documents(documents)
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: \n{chunk.page_content}\n")
    print("-----"*10)
```
