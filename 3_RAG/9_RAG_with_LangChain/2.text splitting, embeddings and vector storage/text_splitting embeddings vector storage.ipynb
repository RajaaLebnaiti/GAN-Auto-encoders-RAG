# 1. Splitting/ chunking

<pre>
- Chunk_size: 
    . too slow: difficult to interpret
    . too big: lack context

- Chunk_overlap: include informations beyond the boundry

### a.) ChracterTextSplitter

```python
from langchain_text_splitters import CharacterTextSplitter

text = """Lorem Ipsum is simply dummy text of the printing and typesetting industry.\n\n
Lorem Ipsum has been the industry's standard dummy text ever since 1966, 
\n\nIt has survived not only many decades, but also the leap into electronic typesetting, 
\n\nIt was popularised thanks to these sheets and more 
recently with desktop publishing software like Aldus PageMaker and Microsoft Word \n\n
including versions of Lorem Ipsum.

 """

# initialiser le CharacterTextSplitter
text_splitter = CharacterTextSplitter(separator= "\n\n",
                                    chunk_size= 100,
                                    chunk_overlap= 10)

```

```python
chunks = text_splitter.split_text(text)
```

```python
chunks
```

```python
print([len(chunk) for chunk in chunks])
```

- chunks may lack context
- chunks may be larger than chunk_size specified 133>>> 100 !! To solve this we can use :

<b> RecursiveCharacterTextSplitter</b>


### b.) RecursiveCharacterTextSplitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    separators= ["\n\n", "\n", " ", ""],  # provide a list of seperators
    chunk_size= 100,
    chunk_overlap = 10
)


# On fait le split selon le premier seperator: "\n\n" si la taille <= chunk_size c'est bon, sinon:
# On fait le split selon le deuxième seperator: "\n" si la taille <= chunk_size c'est bon, sinon:
# On fait le split selon le troisième seperator: " " si la taille <= chunk_size c'est bon, sinon:
# # On fait le split selon le dernier seperator: ""
chunks = splitter.split_text(text)
chunks
```

```python
print([len(chunk) for chunk in chunks]) # Now all the chunks will have a size <= chunk_size
```

### c.) Splitting documents

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("pdf_file_example.pdf")
documents = loader.load()

splitter = RecursiveCharacterTextSplitter(
    chunk_size = 1000,
    chunk_overlap= 200
)


chunks = splitter.split_documents(documents)

chunks
```

```python
print([len(chunk.page_content) for chunk in chunks])
```
