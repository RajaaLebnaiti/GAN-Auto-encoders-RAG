## Dense retrieval vs Sparse retrieval

### Dense:

- Encode chunks as a single vector with non-zero components.(Toutes les valeurs sont non nulles.) 
- It exceles in capturing semantic meaning, But it's computationally expensive.

### Sparse:
 - Used in BM25 or TF-IDF
 - Encode using word matching with mostly zero components.(Le vecteur contient presque uniquement des 0, sauf aux positions des mots présents dans le texte)
 - It is precise, explainable and good in handling rare-words, But it is less generalizable.

```python
from langchain_community.retrievers import BM25Retriever

chunks = [
    "Python was created by Guido van Rossum and released in 1991.",
    "Python is a popular language for machine learning (ML).",
    "The PyTorch library is a popular Python library for AI and ML.",
    "Python was created in 1991 by Guido."
]

bm25_retriever = BM25Retriever.from_texts(chunks, k=3) 
#k=3 est le nombre maximum de documents (ou chunks) que le retriever doit renvoyer 
# après avoir calculé la pertinence.
```

```python
results = bm25_retriever.invoke("When was Python created?")
print("Les résultats : \n")
print(results)


print("\n\nMost relevant Document:")
print(results[0].page_content)
```

```python
# BM25 with documents in RAG
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser


retriever = BM25Retriever.from_documents(
    documents= chunks,
    k= 5
)





prompt = "....."
llm = "....."
# integrate the retriever in the LCEL Chain

chain = (
    {"context": retriever, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
    )


print(chain.invoke("How can LLM hallucination impact a RAG application?"))
```
