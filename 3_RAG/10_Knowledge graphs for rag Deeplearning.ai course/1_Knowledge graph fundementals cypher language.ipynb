# Lesson 2: Querying Knowledge Graphs with Cypher

### Import packages and set up Neo4

```python
from dotenv import load_dotenv
import os

from langchain_community.graphs import Neo4jGraph
```

```python
load_dotenv('.env', override=True)
NEO4J_URI = os.getenv('NEO4J_URI')
NEO4J_USERNAME = os.getenv('NEO4J_USERNAME')
NEO4J_PASSWORD = os.getenv('NEO4J_PASSWORD')
NEO4J_DATABASE = os.getenv('NEO4J_DATABASE')
```

Initialize a knowledge graph instance using LangChain's Neoj4 integration


```python
kg = Neo4jGraph(
    url= NEO4J_URI,
    username= NEO4J_USERNAME,
    password= NEO4J_PASSWORD,
    database= NEO4J_DATABASE
)
```

### Querying the movie knowledge graph 
- Match all nodes in the graph

```python
cypher = """
    MATCH (n)
    RETURN count(n) 
"""
# RETOURNE LE nombre de noeuds dans le graphe
result = kg.query(cypher)
result
```

```python
# ajouter un alias
cypher= """ 
MATCH (n)
RETURN count(n) as numberofNodes
"""

result= kg.query(cypher)
result
```

```python
print(f"There are {result[0]['numberofNodes']} nodes in this graph.")
```

```python
# récupérer d'un node spécifique : Person
cypher= """ 
MATCH (person: Person)
RETURN count(person) as numberofNodesPerson
"""

result= kg.query(cypher)
print(f"The number of \'Person\' nodes is : {result[0]['numberofNodesPerson']}")
```

```python
# récupérer d'un node spécifique : Location
cypher= """ 
MATCH (loc: Location)
RETURN count(loc) as numberofNodesLocation
"""

result= kg.query(cypher)
print(f"The number of \'Location\' nodes is : {result[0]['numberofNodesLocation']}")
```

```python
# récupérer d'un node spécifique et une valeur spécifique
cypher= """ 
MATCH (loc: Location {id: "San Francisco"})
RETURN loc
"""

result= kg.query(cypher)
print(f"The number of \'Location\' nodes is : {result[0]}")
```

```python
# Cypher patterns with conditional matching

cypher= """ 
MATCH (nineties:Movie)
WHERE nineties.released >= 1900
AND nineties < 2000
RETURN nineties.title
"""
result = kg.query(cypher)
print(result)
```

```python
# adding data to graph

cypher = """ 

CREATE (rajaa: Person {name: "RajaaL"})
RETURN rajaa"""

kg.query(cypher)
```

```python
# Deleted node

cypher = """
    MATCH (p: Person {name: "RajaaL"})
    DELETE p"""
kg.query(cypher)
```
